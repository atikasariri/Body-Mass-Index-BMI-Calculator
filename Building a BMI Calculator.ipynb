# BMI Calculator
https://www.cdc.gov/bmi/adult-calculator/

```python
name = input("Enter your name: ")
weight = int(input("Enter your weight in pounds: "))

height = int(input("Enter your weight in inches: "))
               
BMI = (weight * 703) / (height * height)

print(BMI)

if BMI>0:
    if(BMI<18.5):
        print(name +",You are underweight. You need to eat more and stop prouding flat body.")
    elif (BMI<=24.9):
        print(name +",You are healthyweight.")
    elif (BMI<29.9):
        print(name +",You are Overweight.")
    else:
        print(name +",You are Obesity.")
else:
    print("Enter valid input")
```
